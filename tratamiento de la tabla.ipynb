```python
query = """
    SELECT
        duration_sec,
        viral
    FROM viral_videos
"""
df_sec = pd.read_sql(query, con= engine)
```

```python
maximo_duracion = df_sec['duration_sec'].max()
```

```python
numero_de_grupos = 50
grupos = [lambda x: (maximo_duracion/numero_de_grupos) * x for x in range(0, numero_de_grupos + 1)]
df_sec['duration_sec'] = pd.cut(df_sec['duration_sec'], bin = grupos)
```
